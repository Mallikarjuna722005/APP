# Scenario 4: Online Food Delivery System

## Objective
Automatically assign the nearest available delivery partner while following workload, rating, distance, and idle-time rules.

## Business Rules
1. Only available delivery partners are eligible.
2. Select the partner with the fewest deliveries today.
3. If tied, select the partner with the highest rating.
4. If still tied, select the nearest partner.
5. If still tied, select the partner idle for the longest time.
6. After assignment, update the partner status to Busy.
7. Increment deliveries completed today.

## Input
Restaurant Location: (6, 7)

D101 - Rahul - (12, 11) - Available - 8 deliveries - Rating 5  
D102 - Anu - (9, 8) - Available - 4 deliveries - Rating 4  
D103 - Kiran - (9, 8) - Busy - 4 deliveries - Rating 3  
D104 - John - (11, 9) - Available - 2 deliveries - Rating 5

## Expected Output
Delivery Assigned to D102

```python
partners = [
    {"id": "D101", "name": "Rahul", "location": (12, 11),
     "status": 1, "deliveries": 8, "rating": 5, "idle": 10},

    {"id": "D102", "name": "Anu", "location": (9, 8),
     "status": 1, "deliveries": 4, "rating": 4, "idle": 15},

    {"id": "D103", "name": "Kiran", "location": (9, 8),
     "status": 0, "deliveries": 4, "rating": 3, "idle": 20},

    {"id": "D104", "name": "John", "location": (11, 9),
     "status": 1, "deliveries": 2, "rating": 5, "idle": 5}
]

restaurant = (6, 7)

available = []

for partner in partners:
    if partner["status"] == 1:
        available.append(partner)

if available:
    selected = available[0]

    for partner in available[1:]:
        if partner["deliveries"] < selected["deliveries"]:
            selected = partner

        elif partner["deliveries"] == selected["deliveries"]:
            if partner["rating"] > selected["rating"]:
                selected = partner

            elif partner["rating"] == selected["rating"]:
                distance1 = ((selected["location"][0] - restaurant[0]) ** 2 +
                             (selected["location"][1] - restaurant[1]) ** 2)

                distance2 = ((partner["location"][0] - restaurant[0]) ** 2 +
                             (partner["location"][1] - restaurant[1]) ** 2)

                if distance2 < distance1:
                    selected = partner

                elif distance2 == distance1:
                    if partner["idle"] > selected["idle"]:
                        selected = partner

    selected["status"] = 0
    selected["deliveries"] += 1

    print("Delivery Assigned to", selected["id"])

else:
    print("All delivery partners are busy.")
```
