```python
#Scenario 1: Student Marks Management

class InvalidMarksError(Exception):
    pass

while True:
    try:
        name = input("Enter student name: ")
        assert name.strip() != "", "Name cannot be empty"

        marks = float(input("Enter marks: "))

        if marks < 0 or marks > 100:
            raise InvalidMarksError("Marks must be between 0 and 100")

        with open("student_records.txt", "a") as file:
            file.write(f"{name},{marks}\n")

        print("Record saved")

    except AssertionError as e:
        print(e)
    except ValueError:
        print("Marks must be numeric")
    except InvalidMarksError as e:
        print(e)
    finally:
        print("Processing completed")

    choice = input("Enter another record? (yes/no): ")
    if choice.lower() != "yes":
        break
```

```python
# Scenario 2: NumPy Array Insertion

import numpy as np

arr = np.array([10, 20, 30, 40, 50])

try:
    print("Original array:", arr)

    position = int(input("Enter position: "))
    element = input("Enter new element: ")

    if not element.isdigit():
        raise ValueError("Element must be an integer")

    if position < 0 or position > len(arr):
        raise IndexError("Invalid position")

    arr = np.insert(arr, position, int(element))

    print("Updated array:", arr)

except ValueError as e:
    print(e)
except IndexError as e:
    print(e)
finally:
    print("Processing completed")
```

```python
# Scenario 3: Precious Metal Loan Management
def calculate_interest(metal, purity, amount, tenure):
    rates = {
        "gold": {18: 8, 22: 7, 24: 6},
        "silver": {925: 9, 950: 8, 999: 7},
        "platinum": {900: 7, 950: 6, 999: 5}
    }

    metal = metal.lower()

    if metal not in rates:
        raise ValueError("Invalid metal")

    if purity not in rates[metal]:
        raise ValueError("Invalid purity")

    rate = rates[metal][purity]
    return amount * rate * tenure / 100


try:
    metal = input("Enter metal: ")
    purity = int(input("Enter purity: "))
    amount = float(input("Enter loan amount: "))
    tenure = int(input("Enter tenure: "))

    interest = calculate_interest(metal, purity, amount, tenure)

except TypeError:
    print("Missing required argument")
except ValueError as e:
    print(e)
else:
    print("Interest:", interest)
finally:
    print("Loan processing completed")
```
